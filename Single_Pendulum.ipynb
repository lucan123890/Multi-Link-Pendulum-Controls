```python
# import sympy for symbolic computations
from sympy import *
# latex math display
from IPython.display import Javascript, Math, HTML

def latexDisplay(latex_obj):
  display(Javascript("var sc = document.createElement('script')")); display(Javascript("sc.type='text/javascript'; sc.src='https://cdnjs.cloudflare.com/ajax/libs/mathjax/2.7.3/MathJax.js?config=TeX-AMS-MML_HTMLorMML&dummy=.js'")); display(Javascript("var currentDiv = document.activeElement.parent")); display(Javascript("document.body.insertBefore(sc, currentDiv)"))
  display(Math(latex_obj));
```

```python
#Variable definitions
#generalized coordinates and their derivatives
q0 = Symbol('q_0', real=True); # cart translational position (redefined from x_c in figure)
q1 = Symbol('q_1', real=True); # first joint rotational position
q = Matrix([q0, q1]);

dq0 = Symbol('\dot{q}_0', real=True); # cart translational velocity
dq1 = Symbol('\dot{q}_1', real=True); # first joint rotational velocity
dq = Matrix([dq0, dq1]);

ddq0 = Symbol('\ddot{q}_0', real=True); # cart translational acceleration
ddq1 = Symbol('\ddot{q}_1', real=True); # first joint rotational acceleration
ddq = Matrix([ddq0, ddq1]);

# input - force applied to cart
f = Symbol('f', real=True);
```

```python
# kinematic parameter
l1, l2 = symbols('l_1, l_2', real=True, positive=True); # length of each link
a1, a2, a3  = symbols('a_1, a_2, a_3', real=True, positive=True); # distance to the center of mass of each link from its pivot (redefined from r_1, r_2, r_3 in figure)

# dynamic parameter
J1, J2, J3 = symbols('J_1, J_2, J_3', real=True, positive=True); # moment of inertia of each link
m0, m1, m2, m3 = symbols('m_0, m_1, m_2, m_3', real=True, positive=True); # mass of each link
d0, d1, d2, d3 = symbols('d_0, d_1, d_2, d_3', real=True, positive=True); # relative friction of each link

grav = symbols('g', real=True, positive=True); # garvitational constant
```

```python
# time dependency and dictionary for substiution
t = symbols('t', real=True, positive=True); # time

# positions
q0_t = Function('q_0', real=True)(t); # cart translational position
q1_t = Function('q_1', real=True)(t); # first joint rotational position
q_t = Matrix([q0_t, q1_t]);

q_t2s = {q0_t:q0, q1_t:q1}; # qx(t) -> qx
q_s2t = {q0:q0_t, q1:q1_t}; # qx -> qx(t)

# velocities
dq0_t = Function('\dot{q}_0', real=True)(t); # cart translational velocity
dq1_t = Function('\dot{q}_1', real=True)(t); # first joint rotational velocity
dq_t = Matrix([dq0_t, dq1_t]);

dq_t2s = {dq0_t:dq0, dq1_t:dq1}; # dqx(t) -> dqx
dq_s2t = {dq0:dq0_t, dq1:dq1_t}; # dqx -> dqx(t)

q2dq = {diff(q0_t,t):dq0_t, diff(q1_t,t):dq1_t}; # substitute first derivatives

# accelerations
ddq0_t = Function('\ddot{q}_0', real=True)(t); # cart translational acceleration
ddq1_t = Function('\ddot{q}_1', real=True)(t); # first joint rotational acceleration
ddq_t = Matrix([ddq0_t, ddq1_t]);

ddq_t2s = {ddq0_t:ddq0, ddq1_t:ddq1}; # ddqx(t) -> ddqx
ddq_s2t = {ddq0:ddq0_t, ddq1:ddq1_t}; # ddqx -> ddqx(t)

dq2ddq = {diff(dq0_t,t):ddq0_t, diff(dq1_t,t):ddq1_t}; # substitute second derivatives
```

```python
# planar rotation
def rot(q):
    return Matrix([[cos(q), -sin(q)],[sin(q), cos(q)]]);
```

```python
# positions of center of mass
p0 = Matrix([q0_t, 0]);
p1 = p0 + rot(q1_t)*Matrix([0, a1]);

latexDisplay(r'p_0 =' + latex(p0.subs(q_t2s)));
latexDisplay(r'p_1 =' + latex(p1.subs(q_t2s)));
```

```python
# velocity of center of mass
dp0 = diff(p0, t).subs(q2dq);
dp1 = diff(p1, t).subs(q2dq);

latexDisplay(r'\dot{p}_0 =' + latex(dp0.subs(q_t2s).subs(dq_t2s)));
latexDisplay(r'\dot{p}_1 =' + latex(dp1.subs(q_t2s).subs(dq_t2s)));
```

```python
#Defining Lagrangian
#mcart = 1
#m1 = 0.08625
#l = 0.1778
#a = 0.12991
#b = 2.2394 / 10**4
#J1 = 0.000473883

KE = Rational(1,2)*m0*dp0.T*dp0 + Rational(1,2)*m1*dp1.T*dp1 + (Rational(1,2)*J1*dq1_t**2 * eye(1)) #Ev_cart + Ev_link + Ew_link
PE = m0*Matrix([0, -grav]).T*p0 + m1*Matrix([0, -grav]).T*p1

latexDisplay(r'KE =' + latex(KE.subs(q_t2s).subs(dq_t2s)));
latexDisplay(r'PE =' + latex(PE.subs(q_t2s).subs(dq_t2s)));

```

```python
# Lagrangian
L = KE-PE;

# Lagrange equation
Lode = diff(Matrix([L]).jacobian(dq_t),t) - Matrix([L]).jacobian(q_t); Lode = Lode.T;
Lode = Lode.subs(q2dq).subs(dq2ddq);

latexDisplay(r'EOM: 0 =' + latex(Lode.subs(q_t2s).subs(dq_t2s).subs(ddq_t2s)));
```

```python
#linearize
#Lode = Lode.subs({q0_t:0, q1_t:0, dq0_t:0, dq1_t:0})
latexDisplay(r'0 =' + latex(simplify(Lode.subs(q_t2s).subs(dq_t2s).subs(ddq_t2s))));
```

```python
#Inlude damping on the pendulum swing
D = Matrix([[0,0],[0,d1]])

latexDisplay(r'0 =' + latex(D*dq));
Lode = Lode - D*dq
latexDisplay(r'0 =' + latex(simplify(Lode.subs(q_t2s).subs(dq_t2s).subs(ddq_t2s))));
```

```python
#Small-angle/equilibrium linearization substitutions
lin_subs = {
    sin(q1_t): q1_t,
    cos(q1_t): 1,
    dq1_t**2: 0,
    dq0_t**2: 0,
    dq0_t*dq1_t: 0,
    dq1_t*dq0_t: 0,
    q1_t**2: 0
}

#Apply substitution to Lode (the equations of motion)
Lode_lin = Lode.subs(lin_subs)

#Also substitute equilibrium values for velocities and positions
eq_subs = {
    q0_t: 0,
    q1_t: 0,
    dq0_t: 0,
    dq1_t: 0
}
```

```python
#find mass matrix to calculate the velocity derivatives in 'A' matrix
M = simplify(Lode_lin.jacobian(ddq_t).subs(eq_subs))
latexDisplay(r'M='+latex(simplify(M)))
Minv = M**-1

latexDisplay(r'M^{-1}='+latex(Minv))
```

```python
#Rephrase to view the reset of the EOM terms
Lode_remaining = Lode_lin - M*ddq_t; Lode_remaining = simplify(Lode_remaining)
latexDisplay(r'0=Mddq+f(q)')
latexDisplay(r'0=' + latex(M)+latex(ddq)+r'+'+latex(Lode_remaining.subs(q_t2s).subs(dq_t2s).subs(ddq_t2s)))
```

```python
#Get state space equations for dq
latexDisplay(r'ddq/dt^2=M^{-1}(-f(q))')
latexDisplay(latex(ddq)+r'='+latex(Minv*Lode_remaining.subs(q_t2s).subs(dq_t2s).subs(ddq_t2s)))

#B matrix bottom half
latexDisplay(r'B=M^{-1}'+latex(Matrix([1,0])))
latexDisplay(r'B='+latex(Minv*Matrix([1,0])))
```

```python
#Convert to state-space equation
states = Matrix([q0,q1,dq0,dq1])
state_derivative = Matrix([dq0,dq1,ddq0,ddq1])

denom = J1*m0 + J1*m1 + a1**2*m0*m1

#linearizing by hand
A = Matrix([[0,0,1,0],
            [0,0,0,1],
            [0,(a1**2*m1**2*grav)/denom,0,-(d1*a1*m1)/denom],
            [0,(a1*m1*grav*(m0+m1))/denom,0,-(d1*(m0+m1))/denom]])

B = Matrix([0,0,(J1+a1**2*m1)/denom,(a1*m1)/denom])

latexDisplay(r'd/dt'+latex(states) + r'=' + latex(state_derivative) + r'=' + latex(A)+latex(states)+ r'+' + latex(B)+r'u');
```

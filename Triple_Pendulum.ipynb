```python
#Variable definitions
#generalized coordinates and their derivatives
q0 = Symbol('q_0', real=True); # cart translational position (redefined from x_c in figure)
q1 = Symbol('q_1', real=True); # first joint rotational position
q2 = Symbol('q_2', real=True); # second joint rotational position
q3 = Symbol('q_3', real=True); # third joint rotational position
q = Matrix([q0, q1, q2, q3]);

dq0 = Symbol('\dot{q}_0', real=True); # cart translational velocity
dq1 = Symbol('\dot{q}_1', real=True); # first joint rotational velocity
dq2 = Symbol('\dot{q}_2', real=True); # second joint rotational velocity
dq3 = Symbol('\dot{q}_3', real=True); # third joint rotational velocity
dq = Matrix([dq0, dq1, dq2, dq3]);

ddq0 = Symbol('\ddot{q}_0', real=True); # cart translational acceleration
ddq1 = Symbol('\ddot{q}_1', real=True); # first joint rotational acceleration
ddq2 = Symbol('\ddot{q}_2', real=True); # second joint rotational acceleration
ddq3 = Symbol('\ddot{q}_3', real=True); # third joint rotational acceleration
ddq = Matrix([ddq0, ddq1, ddq2, ddq3]);

# input - force applied to cart
f = Symbol('f', real=True);
```

```python
# kinematic parameter
l1, l2 = symbols('l_1, l_2', real=True, positive=True); # length of each link
a1, a2, a3  = symbols('a_1, a_2, a_3', real=True, positive=True); # distance to the center of mass of each link from its pivot (redefined from r_1, r_2, r_3 in figure)

# dynamic parameter
J1, J2, J3 = symbols('J_1, J_2, J_3', real=True, positive=True); # moment of inertia of each link
m0, m1, m2, m3 = symbols('m_0, m_1, m_2, m_3', real=True, positive=True); # mass of each link
d0, d1, d2, d3 = symbols('d_0, d_1, d_2, d_3', real=True, positive=True); # relative friction of each link

grav = symbols('g', real=True, positive=True); # garvitational constant
```

```python
# time dependency and dictionary for substiution
t = symbols('t', real=True, positive=True); # time

# positions
q0_t = Function('q_0', real=True)(t); # cart translational position
q1_t = Function('q_1', real=True)(t); # first joint rotational position
q2_t = Function('q_2', real=True)(t); # second joint rotational position
q3_t = Function('q_3', real=True)(t); # third joint rotational position
q_t = Matrix([q0_t, q1_t, q2_t, q3_t]);

q_t2s = {q0_t:q0, q1_t:q1, q2_t:q2, q3_t:q3}; # qx(t) -> qx
q_s2t = {q0:q0_t, q1:q1_t, q2:q2_t, q3:q3_t}; # qx -> qx(t)

# velocities
dq0_t = Function('\dot{q}_0', real=True)(t); # cart translational velocity
dq1_t = Function('\dot{q}_1', real=True)(t); # first joint rotational velocity
dq2_t = Function('\dot{q}_2', real=True)(t); # second joint rotational velocity
dq3_t = Function('\dot{q}_3', real=True)(t); #third joint rotational velocity
dq_t = Matrix([dq0_t, dq1_t, dq2_t, dq3_t]);

dq_t2s = {dq0_t:dq0, dq1_t:dq1, dq2_t:dq2, dq3_t:dq3}; # dqx(t) -> dqx
dq_s2t = {dq0:dq0_t, dq1:dq1_t, dq2:dq2_t, dq3:dq3_t}; # dqx -> dqx(t)

q2dq = {diff(q0_t,t):dq0_t, diff(q1_t,t):dq1_t, diff(q2_t,t):dq2_t, diff(q3_t,t):dq3_t}; # substitute first derivatives

# accelerations
ddq0_t = Function('\ddot{q}_0', real=True)(t); # cart translational acceleration
ddq1_t = Function('\ddot{q}_1', real=True)(t); # first joint rotational acceleration
ddq2_t = Function('\ddot{q}_2', real=True)(t); # second joint rotational acceleration
ddq3_t = Function('\ddot{q}_3', real=True)(t); # third joint rotational acceleration
ddq_t = Matrix([ddq0_t, ddq1_t, ddq2_t, ddq3_t]);

ddq_t2s = {ddq0_t:ddq0, ddq1_t:ddq1, ddq2_t:ddq2, ddq3_t:ddq3}; # ddqx(t) -> ddqx
ddq_s2t = {ddq0:ddq0_t, ddq1:ddq1_t, ddq2:ddq2_t, ddq3:ddq3_t}; # ddqx -> ddqx(t)

dq2ddq = {diff(dq0_t,t):ddq0_t, diff(dq1_t,t):ddq1_t, diff(dq2_t,t):ddq2_t, diff(dq3_t,t):ddq3_t}; # substitute second derivatives
```

```python
# planar rotation
def rot(q):
    return Matrix([[cos(q), -sin(q)],[sin(q), cos(q)]]);
```

```python
# positions of center of mass
p0 = Matrix([q0_t, 0]);
p1 = p0 + rot(q1_t)*Matrix([0, a1]);
p2 = p0 + rot(q1_t)*Matrix([0, l1]) + rot(q2_t)*Matrix([0, a2]);
p3 = p0 + rot(q1_t)*Matrix([0, l1]) + rot(q2_t)*Matrix([0, l2]) + rot(q3_t)*Matrix([0, a3]);

latexDisplay(r'p_0 =' + latex(p0.subs(q_t2s)));
latexDisplay(r'p_1 =' + latex(p1.subs(q_t2s)));
latexDisplay(r'p_2 =' + latex(p2.subs(q_t2s)));
latexDisplay(r'p_3 =' + latex(p3.subs(q_t2s)));
```

```python
# velocity of center of mass
dp0 = diff(p0, t).subs(q2dq);
dp1 = diff(p1, t).subs(q2dq);
dp2 = diff(p2, t).subs(q2dq);
dp3 = diff(p3, t).subs(q2dq);

latexDisplay(r'\dot{p}_0 =' + latex(dp0.subs(q_t2s).subs(dq_t2s)));
latexDisplay(r'\dot{p}_1 =' + latex(dp1.subs(q_t2s).subs(dq_t2s)));
latexDisplay(r'\dot{p}_2 =' + latex(dp2.subs(q_t2s).subs(dq_t2s)));
latexDisplay(r'\dot{p}_3 =' + latex(dp3.subs(q_t2s).subs(dq_t2s)));
```

```python
#Defining Lagrangian

#Translational Kinetic Energy
KE_T = Rational(1,2)*m0*dp0.T*dp0 + Rational(1,2)*m1*dp1.T*dp1 + Rational(1,2)*m2*dp2.T*dp2 + Rational(1,2)*m3*dp3.T*dp3

#Rotational Kinetic Energy
KE_R = (Rational(1,2)*J1*dq1_t**2 * eye(1)) + (Rational(1,2)*J2*dq2_t**2 * eye(1)) + (Rational(1,2)*J3*dq3_t**2 * eye(1))

KE = KE_T + KE_R
#Potential Energy
PE = m0*Matrix([0, grav]).T*p0 + m1*Matrix([0, grav]).T*p1 + m2*Matrix([0,grav]).T*p2 + m3*Matrix([0,grav]).T*p3

latexDisplay(r'KE =' + latex(KE.subs(q_t2s).subs(dq_t2s)));
latexDisplay(r'PE =' + latex(PE.subs(q_t2s).subs(dq_t2s)));

```

```python
# Lagrangian
L = KE-PE;

# Lagrange equation
Lode = diff(Matrix([L]).jacobian(dq_t),t) - Matrix([L]).jacobian(q_t); Lode = Lode.T;
Lode = Lode.subs(q2dq).subs(dq2ddq);

latexDisplay(r'EOM: 0 =' + latex(simplify(Lode.subs(q_t2s).subs(dq_t2s).subs(ddq_t2s))));
```

```python
#Small-angle/equilibrium linearization substitutions
lin_subs = {
    sin(q1_t): q1_t,
    cos(q1_t): 1,
    sin(q2_t): q2_t,
    cos(q2_t): 1,
    sin(q3_t): q3_t,
    cos(q3_t):1,
    dq3_t**2: 0,
    dq2_t**2: 0,
    dq1_t**2: 0,
    dq0_t**2: 0,
    dq0_t*dq1_t: 0,
    dq1_t*dq0_t: 0,
    dq0_t*dq2_t: 0,
    dq1_t*dq2_t: 0,
    dq2_t*dq1_t: 0,
    dq2_t*dq0_t: 0,
    dq3_t*dq0_t: 0,
    dq3_t*dq1_t: 0,
    dq3_t*dq2_t: 0,
    dq0_t*dq3_t: 0,
    dq1_t*dq3_t: 0,
    dq2_t*dq3_t: 0,
    q1_t**2: 0,
    q2_t**2: 0,
    q3_t**2: 0,
    cos(q2_t - q1_t): 1,
    sin(q1_t - q2_t): q1_t - q2_t,
    cos(q3_t - q2_t): 1,
    sin(q2_t - q3_t): q2_t - q3_t,
    cos(q1_t - q3_t): 1,
    sin(q1_t - q3_t): q1_t - q3_t
}

#substituting equilibrium values for velocities and positions
eq_subs = {
    q0_t: 0,
    q1_t: 0,
    q2_t: 0,
    q3_t: 0,
    dq0_t: 0,
    dq1_t: 0,
    dq2_t: 0,
    dq3_t: 0,
}
#latexDisplay(r'0 =' + latex(simplify(Lode_lin.subs(eq_subs).subs(q_t2s).subs(dq_t2s).subs(ddq_t2s))));
#latexDisplay(r'\quad');
#latexDisplay(r'0 =' + latex(simplify(Lode_lin.subs(q_t2s).subs(dq_t2s).subs(ddq_t2s))));

#Lode_simplified = Lode_lin.subs(eq_subs)
```

```python
#Find Mass matrix
M = simplify(Lode.jacobian(ddq_t)); M = factor(M);
latexDisplay(r'M(\ddot{q})='+latex(M.subs(q_t2s))+latex(ddq))

#Remove mass terms from EOMs to get the remaining terms
Lode_diff = simplify(Lode - M*ddq_t)

```

```python
#find Coriolis matrix and confirm that dM-2C is skew-symmetric
#Skew-symmetry is necessary to ensure conservation of energy and lyapunov stability (I think)
C = simplify(Rational(1,2)*Lode.jacobian(dq_t));
latexDisplay(r'C(q,\dot{q})='+latex(C.subs(q_t2s).subs(dq_t2s)))
latexDisplay(r'\quad');

skew_test = (diff(M, t).subs(q2dq) - 2*C) + (diff(M, t).subs(q2dq) - 2*C).T; skew_test = simplify(simplify(skew_test));
if skew_test.is_zero_matrix:
  latexDisplay(r'C(q, \dot{q}) \text{ defined such that } \dot{M}(q) - 2C(q, \dot{q}) \text{ is skew-symmetric}');
else:
  latexDisplay(r'\text{something went wrong}');
```

```python
g = simplify(Lode_diff - C*dq_t)
latexDisplay(r'g(q) =' + latex(g.subs(q_t2s).subs(dq_t2s)))
```

```python
#check that all terms have been categorized
ode_check = M*ddq_t+C*dq_t+g-Lode; ode_check = simplify(ode_check);
if ode_check.is_zero:
  latexDisplay(r'\text{robotic equation is valid}');

latexDisplay(r'M(q) \ddot{q} + C(q, \dot{q}) \dot{q} + g(q) = -D \dot{q} + \begin{bmatrix} 1 \\ 0 \\ 0 \\ 0 \\ \end{bmatrix} f');
```

```python
#Inlude damping on the pendulum motion
D = Matrix([[0,0,0,0],[0,d1+d2,-d2,0],[0,-d2,d2+d3,-d3],[0,0,-d3,d3]])

latexDisplay(r'D =' + latex(D));
```

```python
#Summary of the system
latexDisplay(r'M='+latex(M.subs(q_t2s)))
latexDisplay(r'C='+latex(C.subs(q_t2s).subs(dq_t2s)))
latexDisplay(r'D='+latex(D.subs(q_t2s)))
latexDisplay(r'g='+latex(g.subs(q_t2s)))
```

```python
# Linearize terms
M_lin = M.subs(lin_subs)
C_lin = C.subs(lin_subs)
g_lin = g.subs(lin_subs)
D_lin = D.subs(lin_subs)
latexDisplay(r'M='+latex(M_lin.subs(q_t2s)))
latexDisplay(r'C='+latex(C_lin.subs(q_t2s).subs(dq_t2s)))
latexDisplay(r'D='+latex(D_lin.subs(q_t2s)))
latexDisplay(r'g='+latex(g_lin.subs(q_t2s)))
```

```python
#Finding A and B matrices

#find RHS in form f(x,u)
RHS_lin = -D_lin*dq_t - g_lin - C_lin*dq_t + Matrix([1,0,0,0])*f
ddq_full = simplify(M_lin.inv() * RHS_lin)
latexDisplay(latex(Matrix([ddq0,ddq1,ddq2,ddq3])) + r'=' + latex(ddq_full))

#define the states for state space equation
x = Matrix.vstack(q_t,dq_t)
latexDisplay(r'x=' + latex(x.subs(q_t2s).subs(dq_t2s)))

#RHS including dq terms
f_xu = Matrix.vstack(dq_t,ddq_full)

#f(x,u) = Ax + Bu
A = f_xu.jacobian(x).subs(eq_subs)
B = f_xu.jacobian(Matrix([f])).subs(eq_subs)

latexDisplay(r'\dot{x}=Ax + Bu')
latexDisplay(latex(Matrix.vstack(dq,ddq)) + r'=' + latex(simplify(A)) + latex(x) + r'+' + latex(simplify(B)) + r'f')
```
